# W2-D1: Alert Correlation - Tu Noise Sang Signal

Notebook nay load alert sample + service topology, sau do chay correlation pipeline gom 20 alert thanh cac cluster co y nghia.


```python
from pathlib import Path
import json
from correlate import load_alerts, load_services, correlate

BASE = Path('.')
alerts = load_alerts(BASE / 'dataset' / 'alerts_sample.jsonl')
services_doc = load_services(BASE / 'dataset' / 'services.json')
print(f"alerts={len(alerts)} services={len(services_doc['services'])} stores={len(services_doc['stores'])}")
```

```python
GAP_SEC = 120
MAX_HOP = 1
summary = correlate(alerts, services_doc, gap_sec=GAP_SEC, max_hop=MAX_HOP)
print(f"gap_sec={GAP_SEC} max_hop={MAX_HOP} output_clusters={summary['output_clusters']} reduction_ratio={summary['reduction_ratio']}")
```

```python
[{k: c[k] for k in ['cluster_id', 'alert_count', 'services', 'max_severity']} for c in summary['clusters']]
```

```python
output_path = BASE / 'results' / 'cluster_summary.json'
output_path.parent.mkdir(parents=True, exist_ok=True)
output_path.write_text(json.dumps(summary, indent=2, ensure_ascii=False) + "\n", encoding="utf-8")
print(f"wrote {output_path}")
```
