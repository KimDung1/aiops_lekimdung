# W2-D2: RCA - Graph, Causal & Retrieval-Augmented

Notebook nay phan tich cluster tu W2-D1, rank root-cause candidates bang graph + temporal scoring, sau do enrich bang incident history retrieval.


```python
from pathlib import Path
import json
from rca import load_json, load_alerts, load_history, analyze

BASE = Path('.')
cluster_summary = load_json(BASE.parent / 'd1' / 'results' / 'cluster_summary.json')
alerts = load_alerts(BASE / 'dataset' / 'alerts_sample.jsonl')
services_doc = load_json(BASE / 'dataset' / 'services.json')
history = load_history(BASE / 'dataset' / 'incidents_history.json')
print(f"clusters={cluster_summary['output_clusters']} alerts={len(alerts)} history={len(history)}")
```

```python
rca_output = analyze(cluster_summary, alerts, services_doc, history)
print(f"clusters_analyzed={rca_output['clusters_analyzed']}")
```

```python
[{k: r[k] for k in ['cluster_id', 'root_cause', 'class', 'confidence']} for r in rca_output['results']]
```

```python
output_path = BASE / 'results' / 'rca_output.json'
output_path.parent.mkdir(parents=True, exist_ok=True)
output_path.write_text(json.dumps(rca_output, indent=2, ensure_ascii=False) + "\n", encoding="utf-8")
print(f"wrote {output_path}")
```
